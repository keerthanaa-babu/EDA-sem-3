```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('/content/powerconsumption.csv')
```

```python
bins = [50,60,70,80,90]
```

```python
category = pd.cut(df['Humidity'], bins)
category
```

```python
category.value_counts()
```

```python
category.head(70)
```

```python
Humidity_labels=['Low Humidity','Moderate Humidity','High Humidity','Very High Humidity']
df['Humidity_Level']=pd.cut(df['Humidity'],bins,labels=Humidity_labels)
df[['Humidity','Humidity_Level']].tail(30)
```

```python
df['Humidity_Level'].value_counts()
```

```python
df['Humidy_freq']=pd.qcut(df['Humidity'],q=4)
df[['Humidity','Humidy_freq']].tail(30)
df['Humidy_freq'].value_counts()
```

```python
df['Humidity_Equal_Width'] = pd.cut(df['Humidity'], bins=4, precision=1)
print(df['Humidity_Equal_Width'].value_counts())
```

```python
pd.cut(np.random.rand(80), 5, precision=2)
```

```python
randomNumbers = np.random.rand(2000)
category3 = pd.qcut(randomNumbers, 4)
print(category3.value_counts())
```

```python
mean_val = df['Humidity'].mean()
median_val = df['Humidity'].median()

print(f"Mean:   {mean_val:.2f}")
print(f"Median: {median_val:.2f}")
```
