```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/Iris.csv")
```

```python
df.groupby("Species").mean()
```

```python
plt.figure(figsize=(6, 4))
sns.histplot(data=df, x="PetalLengthCm", hue="Species", kde=True, multiple="stack")
plt.title("Distribution of Petal Length by Species")
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.scatterplot(data=df, x="SepalLengthCm", y="SepalWidthCm", hue="Species")
plt.title("Sepal Length vs Sepal Width")
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.boxplot(data=df, x="Species", y="PetalWidthCm")
plt.title("Petal Width Variations Across Species")
plt.show()
```

```python
sns.pairplot(df, hue="Species", height=2.0)
plt.show()
```

```python
numeric_df = df.drop(columns=['Id', 'Species'])
correlation_data = numeric_df.corr()

plt.figure(figsize=(6, 4.5))
sns.heatmap(correlation_data, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)

plt.title("Feature Correlation Heatgraph")
plt.tight_layout()
plt.show()
```
